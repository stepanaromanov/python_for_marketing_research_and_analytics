## SQL / Data Layer

```sql
-- ============================================================
-- Marketing Mix Model — Feature Pipeline
-- Produces one row per week with:
--   • Per-channel geometric adstock (channel-specific decay)
--   • External regressors (competitor, macro, holiday)
--   • Fourier seasonality terms (52w and 26w periods)
-- ============================================================

WITH channel_params AS (
  -- Decay rates must match GT_ADSTOCK in params.yaml / config.py.
  -- Decay = fraction of last week's adstocked spend that carries over.
  -- TV 0.70: ~3-week half-life (brand awareness, slow decay)
  -- Search 0.20: ~0.4-week half-life (direct response, nearly no carryover)
  -- Display 0.50: ~1-week half-life
  -- Social 0.35: ~0.7-week half-life
  -- Email 0.15: minimal carryover (one-shot send)
  SELECT 'tv'      AS channel, 0.70 AS decay UNION ALL
  SELECT 'search',              0.20           UNION ALL
  SELECT 'display',             0.50           UNION ALL
  SELECT 'social',              0.35           UNION ALL
  SELECT 'email',               0.15
),

adstocked AS (
  -- Geometric adstock formula:
  --   adstock(t) = Σ spend(t-k) × decay^k   for k = 0..12
  --
  -- Implemented as a weighted window SUM where the weight of each
  -- historical row is decay^(weeks_ago). ROWS BETWEEN 12 PRECEDING
  -- caps history at 12 weeks — beyond that the weight is < 1.4% for
  -- the fastest-decaying channel (TV), so truncation error is negligible.
  --
  -- NOTE: spend_week must be a row-level column in weekly_media_spend
  -- representing the week that spend was originally placed (for the
  -- DATEDIFF exponent). If your table is already one row per week,
  -- spend_week = week and this simplifies to decay^k via the window offset.
  SELECT
    s.week,
    s.channel,
    SUM(
      s.spend * POW(p.decay, DATEDIFF('week', s.spend_week, s.week))
    ) OVER (
      PARTITION BY s.channel   -- independent carryover per channel
      ORDER BY s.week
      ROWS BETWEEN 12 PRECEDING AND CURRENT ROW
    ) AS adstock_spend
  FROM weekly_media_spend s
  -- JOIN ensures each channel row picks up its own decay rate,
  -- replacing the original hardcoded 0.7 that applied to all channels.
  JOIN channel_params p ON s.channel = p.channel
),

adstocked_wide AS (
  -- Pivot from long format (one row per channel per week)
  -- to wide format (one row per week, one column per channel).
  -- Required because the model expects a single feature row per week.
  -- MAX() is a no-op aggregation here — each CASE branch is non-NULL
  -- for exactly one channel, so MAX picks the only non-NULL value.
  SELECT
    week,
    MAX(CASE WHEN channel = 'tv'      THEN adstock_spend END) AS tv_adstock,
    MAX(CASE WHEN channel = 'search'  THEN adstock_spend END) AS search_adstock,
    MAX(CASE WHEN channel = 'display' THEN adstock_spend END) AS display_adstock,
    MAX(CASE WHEN channel = 'social'  THEN adstock_spend END) AS social_adstock,
    MAX(CASE WHEN channel = 'email'   THEN adstock_spend END) AS email_adstock
  FROM adstocked
  GROUP BY week
),

features AS (
  -- Assemble the final feature row per week:
  --   1. Adstocked media spend (from adstocked_wide)
  --   2. External regressors (LEFT JOIN — see note below)
  --   3. Fourier seasonality terms
  --
  -- LEFT JOIN on external_macro:
  --   External sources (competitor GRPs, CPI, rates) update at different
  --   cadences than own media. LEFT JOIN preserves all media weeks even
  --   when external data is delayed. NULLs should be forward-filled
  --   (LOCF) in the Python pipeline for macro series; competitor GRPs
  --   should be flagged as missing rather than imputed.
  --
  -- Fourier terms:
  --   sin/cos pairs encode seasonality without a separate calendar table.
  --   Period 52 = annual cycle; period 26 = semi-annual cycle.
  --   EXTRACT(week FROM ...) returns ISO week number (1–52/53).
  --   Database-specific equivalents:
  --     Snowflake  → WEEKOFYEAR(a.week)
  --     SQL Server → DATEPART(week, a.week)
  --     BigQuery   → EXTRACT(WEEK(MONDAY) FROM a.week)
  SELECT
    a.week,

    -- Adstocked media features (Hill saturation applied later in Python)
    a.tv_adstock,
    a.search_adstock,
    a.display_adstock,
    a.social_adstock,
    a.email_adstock,

    -- External regressors (standardisation applied later in Python)
    e.competitor_tv_grp,   -- competitor brand pressure; expected β < 0
    e.cpi_yoy,             -- inflation index YoY; expected β < 0
    e.rate_delta_bps,      -- interest rate change in bps; expected β < 0
    e.is_holiday_week,     -- binary spike dummy; expected β > 0

    -- Fourier seasonality — period 52 (annual)
    SIN(2 * PI() * EXTRACT(week FROM a.week) / 52) AS sin_52,
    COS(2 * PI() * EXTRACT(week FROM a.week) / 52) AS cos_52,

    -- Fourier seasonality — period 26 (semi-annual)
    SIN(2 * PI() * EXTRACT(week FROM a.week) / 26) AS sin_26,
    COS(2 * PI() * EXTRACT(week FROM a.week) / 26) AS cos_26

  FROM adstocked_wide a
  LEFT JOIN external_macro e ON a.week = e.week
)

-- Final output: one row per week, all features + target.
-- INNER JOIN on weekly_sales intentionally drops weeks with no recorded
-- sales (e.g. pre-launch or data gaps). Use LEFT JOIN + COALESCE(s.sales, 0)
-- if zero-sales weeks should be retained in the training set.
SELECT
  f.*,
  s.sales
FROM features f
JOIN weekly_sales s ON f.week = s.week
ORDER BY f.week;
```

## Realistic Weekly Time-Series Generator

4 years of weekly data with:
- **Own media** (TV, Search, Display, Social, Email) with adstock + Hill saturation
- **External regressors**: competitor TV GRPs, CPI YoY, interest rate delta, holiday dummies
- **Fourier seasonality** terms (52-week and 26-week periods)
- **Injected ground-truth coefficients** for full parameter recovery testing
- **Competitor spend shock** at week 80 for drift detection
- **Series A marketing budget permanent increase** at week 70


````python
# ── Ground-truth parameters — ROAS-anchored DGP ─────────────────────
# Design principle: every parameter is derived from a business target,
# not chosen arbitrarily. ROAS is defined first; betas follow from it.

MEDIA_CHANNELS = ['tv', 'search', 'display', 'social', 'email']
N_WEEKS        = 208   # 4 years
RANDOM_SEED    = 1111
np.random.seed(RANDOM_SEED)
rng = np.random.default_rng(RANDOM_SEED)

# ── Spend levels (weekly average $) ──────────────────────────────────────────
SPEND_BASE = {'tv': 80_000, 'search': 50_000, 'display': 30_000,
              'social': 25_000, 'email': 10_000}

# ── Adstock decay (geometric carryover fraction) ──────────────────────────────
GT_ADSTOCK = {'tv': 0.85, 'search': 0.45, 'display': 0.65,
              'social': 0.55, 'email': 0.35}

# ── Hill curve shape parameters ───────────────────────────────────────────────
# alpha: diminishing returns shape. alpha<1 = concave from origin (TV brand).
#        alpha=1 = hyperbolic. alpha>1 = S-curve with inflection.
# K: half-saturation point in ADSTOCKED spend units.
#    Calibrated so that at mean adstocked spend, Hill output ≈ 0.5–0.7
#    (operating on the slope, not pinned at ceiling or floor).

# Hill saturation: sales_contribution = spend_adstocked^alpha / (spend_adstocked^alpha + K^alpha)
GT_HILL_ALPHA = {'tv': 0.7, 'search': 1.0, 'display': 0.8,
                 'social': 0.9, 'email': 0.8}

GT_HILL_K     = {'tv': 420_000, 'search': 105_000, 'display': 105_000,
                 'social': 80_000, 'email': 18_000}

# ── Target ROAS (business anchor — set these first, derive betas from them) ───
# ROAS = total attributed sales / total spend over full horizon.
# TV brand: low ROAS (slow, diffuse). Search DR: high ROAS (direct intent).
# These are the numbers you would defend to a CMO.
TARGET_ROAS = {'tv': 1.2, 'search': 1.8, 'display': 0.9,
               'social': 1.3, 'email': 2.5}

# ── Baseline (intercept) ─────────────────────────────────────────────────────
# weekly baseline $
GT_BASELINE  = 120_000

# ── External regressor coefficients ──────────────────────────────────────────
# Units: standardised external variable → $ sales impact (same scale as y).
# Sized so macro block ≈ 8–12% of sales variance — realistic for grocery/FMCG.
GT_BETA_EXT = {
    'competitor_tv': -9_000,
    'cpi':           -7_000,
    'rate_delta':    -5_000,
    'holiday':       +18_000,
}

# ── # weekly noise standard deviation ────────────────────────────────────────
GT_NOISE_STD = 18_000 

# ── Channel seasonality phases (week-of-year offset) ─────────────────────────
CHANNEL_PHASES = {'tv': 0, 'search': 13, 'display': 26, 'social': 39, 'email': 6}
CHANNEL_AMP    = {'tv': 0.50, 'search': 0.40, 'display': 0.35,
                  'social': 0.30, 'email': 0.15}

# ── Series A budget step-change ───────────────────────────────────────────────
UPLIFT_FROM_WEEK = 70
SPEND_MULTIPLIER = 1.4 

print('Ground-truth parameters:')
print(f'  Adstock     : {GT_ADSTOCK}')
print(f'  Hill K      : {GT_HILL_K}')
print(f'  Target ROAS : {TARGET_ROAS}')
print(f'  Ext betas   : {GT_BETA_EXT}')
print(f'  Baseline    : ${GT_BASELINE:,}  |  Noise σ: ${GT_NOISE_STD:,}')
````

````python
# ── Transform functions + spend + ROAS-calibrated media betas ────────

def adstock(x: np.ndarray, decay: float) -> np.ndarray:
    """Geometric adstock — carryover: a_t = x_t + decay * a_{t-1}"""
    out = np.zeros_like(x, dtype=float)
    out[0] = x[0]
    for t in range(1, len(x)):
        out[t] = x[t] + decay * out[t - 1]
    return out

def hill(x: np.ndarray, alpha: float, K: float) -> np.ndarray:
    """Hill saturation: x^alpha / (x^alpha + K^alpha), output in [0,1]"""
    xa = np.power(np.clip(x, 0, None), alpha)  # Forces all negative values to zero and raises to the power alpha
    return xa / (xa + K**alpha + 1e-12)   # x<<K→0, x=K→0.5, x>>K→1

def fourier_seasonality(n_weeks: int, periods=(52, 26), order=2) -> pd.DataFrame:
    """Sine/cosine Fourier pairs for each period × harmonic (Fourier seasonality terms for given periods)"""
    t = np.arange(n_weeks)
    cols = {}
    for p in periods:
        for i in range(1, order + 1):
            cols[f'sin_{p}_{i}'] = np.sin(2 * np.pi * i * t / p)
            cols[f'cos_{p}_{i}'] = np.cos(2 * np.pi * i * t / p)
    return pd.DataFrame(cols)

def stdz(x: np.ndarray) -> np.ndarray:
    return (x - x.mean()) / (x.std() + 1e-12)

# ── Weeks index ───────────────────────────────────────────────────────────────
weeks = pd.date_range('2020-01-06', periods=N_WEEKS, freq='W-MON')
t_idx = np.arange(N_WEEKS)

# ── Simulate weekly spend with seasonality + noise ────────────────────────────
media_spend_noise = 0.3

media_spend = pd.DataFrame({
    ch: (np.abs(rng.normal(SPEND_BASE[ch], SPEND_BASE[ch] * media_spend_noise, N_WEEKS))
         * (1.0 + CHANNEL_AMP[ch] * np.sin(2 * np.pi * (t_idx - CHANNEL_PHASES[ch]) / 52)))
    for ch in MEDIA_CHANNELS
}, index=weeks)

# ── Permanent budget uplift from week 70 (investor money) ────────────────
# Rationale: Series A / investor funding received at week 70.
# Marketing budget permanently increased for remainder of campaign.
# 
# Naive OLS: sees higher spend but each week is independent —
#            misses the accumulating carryover stock
# Adstock:   stock builds week-on-week to a NEW higher steady state
#            steady-state adstock = spend / (1 - decay)
# Seasonality: completely blind — week 70 is not a calendar event

# Series A step-change at week 70
for ch in MEDIA_CHANNELS:
    media_spend.iloc[UPLIFT_FROM_WEEK:][ch] *= SPEND_MULTIPLIER
for ch in MEDIA_CHANNELS:
    vals = media_spend[ch].values.copy()
    vals[UPLIFT_FROM_WEEK:] *= SPEND_MULTIPLIER
    media_spend[ch] = vals

# ── Adstock + Hill transformations ────────────────────────────────────────────
media_adstocked    = {}
media_saturated    = {}
for ch in MEDIA_CHANNELS:
    media_adstocked[ch] = adstock(media_spend[ch].values, GT_ADSTOCK[ch])
    media_saturated[ch] = hill(media_adstocked[ch], GT_HILL_ALPHA[ch], GT_HILL_K[ch])

# ── ROAS calibration: derive GT_BETA_MEDIA from TARGET_ROAS ──────────────────
# ROAS = (beta * sum(hill_output)) / sum(spend)
# → beta = TARGET_ROAS * sum(spend) / sum(hill_output)
#
# This guarantees that when you later compute
#   attributed_sales[ch] = GT_BETA_MEDIA[ch] * hill(adstock(spend[ch]))
# and divide by total spend[ch], you get exactly TARGET_ROAS[ch].
GT_BETA_MEDIA = {}
for ch in MEDIA_CHANNELS:
    total_spend    = media_spend[ch].sum()
    total_hill_out = media_saturated[ch].sum()
    GT_BETA_MEDIA[ch] = TARGET_ROAS[ch] * total_spend / total_hill_out

print('ROAS-calibrated media betas:')
for ch in MEDIA_CHANNELS:
    implied_roas = GT_BETA_MEDIA[ch] * media_saturated[ch].sum() / media_spend[ch].sum()
    print(f'  {ch:<8}  beta={GT_BETA_MEDIA[ch]:>10,.0f}  '
          f'implied ROAS={implied_roas:.3f}x  (target={TARGET_ROAS[ch]}x)')

# Sanity: Hill operating point
print('\nHill operating points (mean adstocked spend / K):')
for ch in MEDIA_CHANNELS:
    op = media_adstocked[ch].mean() / GT_HILL_K[ch]
    sat_mean = media_saturated[ch].mean()
    print(f'  {ch:<8}  adstock_mean/K={op:.2f}  hill_mean={sat_mean:.3f}  '
          f'{"OK" if 0.3 < sat_mean < 0.85 else "WARN: near floor/ceiling"}')

````

````python
# ── External regressors + DGP assembly + sanity report ───────────────

# ── External regressors ───────────────────────────────────────────────────────
# Competitor TV — baseline ~100 GRPs/wk, shock at week 80–95
comp_tv          = rng.normal(100, 20, N_WEEKS).clip(0)
comp_tv[80:95]  += 145  # competitor spending spike → drift injection

# CPI YoY % — slow-moving AR(1): 0.92 persistence, slight upward drift
cpi    = np.zeros(N_WEEKS); cpi[0] = 2.5
for t in range(1, N_WEEKS): # Consumer Price Index Year-over-Year inflation rate
    cpi[t] = 0.92 * cpi[t-1] + rng.normal(0.2, 0.3) # AR1 model, 0.92 Autoregressive coefficient 
    # Random shock each week: Mean 0.2 = slight upward drift (inflation tends to slowly increase over long periods) Std 0.3 = moderate volatility
cpi = np.clip(cpi, -1, 12) # Prevents unrealistic values: Inflation cannot go below -1% (rare but possible deflation) 
# or above 12% (extremely high inflation is capped)

# Interest rate delta (bps) — AR(1): 0.80 persistence, faster mean-reversion
rate = np.zeros(N_WEEKS)
for t in range(1, N_WEEKS):
    rate[t] = 0.8 * rate[t-1] + rng.normal(0, 8) # Random shock 1 bps (basis point) = 0.01%
    # Persistence = 0.8: Moderate memory. A rate hike this week increases the probability of another hike next week,
    # but it fades relatively quickly (faster than CPI’s 0.92).
rate = np.clip(rate, -100, 200) # No more than -100 bps cut in one week, No more than +200 bps hike in one week

# Holiday binary dummy — 4 events/year (Christmas×2, Easter, summer)
holiday = np.zeros(N_WEEKS)
for yr in range(4):
    for w in [yr*52+50, yr*52+51, yr*52+15, yr*52+36]:
        if w < N_WEEKS:
            holiday[w] = 1.0

# Standardise continuous external vars (binary holiday left raw)
comp_tv_std = stdz(comp_tv)
cpi_std     = stdz(cpi)
rate_std    = stdz(rate)

# ── Fourier seasonality ───────────────────────────────────────────────────────
fourier      = fourier_seasonality(N_WEEKS, periods=(52, 26), order=2)
season_effect = (0.8 * fourier['sin_52_1'] + 0.5 * fourier['cos_52_1']
               + 0.3 * fourier['sin_26_1'] + 0.2 * fourier['cos_26_1'])
season_sales  = 1_200  * season_effect.values

# ── Sales DGP — all components in $ ──────────────────────────────────────────
# External betas are now in $/std-unit (same scale as y).
# No magic scaling factor — units are explicit throughout.
media_contrib = sum(GT_BETA_MEDIA[ch] * media_saturated[ch] for ch in MEDIA_CHANNELS)
ext_contrib   = (GT_BETA_EXT['competitor_tv'] * comp_tv_std
               + GT_BETA_EXT['cpi']           * cpi_std
               + GT_BETA_EXT['rate_delta']     * rate_std
               + GT_BETA_EXT['holiday']        * holiday)
noise         = rng.normal(0, GT_NOISE_STD, N_WEEKS)

# Unmodelled confounders — real world has these, model doesn't
promo_effect  = np.zeros(N_WEEKS)
promo_weeks   = rng.choice(N_WEEKS, size=18, replace=False)
promo_effect[promo_weeks] = rng.normal(8_000, 3_000, 18)  # flash promos

# slow-moving unobserved trend (brand equity, word-of-mouth)
# Model has no feature for this — pure R² leak
organic_trend = np.cumsum(rng.normal(0, 400, N_WEEKS))
organic_trend -= organic_trend.mean() 

weather_shock = rng.normal(0, 6_000, N_WEEKS)   # uncaptured weather/events

true_sales = (GT_BASELINE
            + media_contrib
            + ext_contrib
            + season_sales
            + organic_trend     # model can't see this  
            + promo_effect      # model can't see this
            + weather_shock     # model can't see this
            + noise)

true_sales = np.clip(true_sales, 0, None)

# ── Master DataFrame ──────────────────────────────────────────────────────────
fourier.index = weeks
df = pd.DataFrame({
    'sales':          true_sales,
    **{f'{ch}_spend': media_spend[ch].values    for ch in MEDIA_CHANNELS},
    **{f'{ch}_sat':   media_saturated[ch]       for ch in MEDIA_CHANNELS},
    'competitor_tv':  comp_tv,   'comp_tv_std': comp_tv_std,
    'cpi':            cpi,       'cpi_std':     cpi_std,
    'rate_delta':     rate,      'rate_std':    rate_std,
    'holiday':        holiday,
    'media_contrib':  media_contrib,
    'ext_contrib_raw':ext_contrib,
    'season_contrib': season_sales,
    'noise_contrib':  noise,
}, index=weeks)
df = pd.concat([df, fourier], axis=1)

EXEC_METRICS = {'n_weeks': N_WEEKS, 'n_media': len(MEDIA_CHANNELS),
                'n_ext': 4, 'drift_week': 80}

# ── Sanity report ─────────────────────────────────────────────────────────────
total_sales = df['sales'].sum()
print('=' * 60)
print('DGP SANITY REPORT')
print('=' * 60)
print(f'Weekly sales   mean=${df.sales.mean():>10,.0f}  std=${df.sales.std():>8,.0f}')
print(f'Noise/signal ratio : {df.noise_contrib.std() / df.media_contrib.std():.2f}x  (target <0.20)')
print()
print(f'{"Channel":<10} {"Avg spend":>12} {"Attr sales":>12} {"ROAS":>8} {"% of sales":>12}')
print('-' * 58)
for ch in MEDIA_CHANNELS:
    attr  = GT_BETA_MEDIA[ch] * media_saturated[ch].sum()
    spend = media_spend[ch].sum()
    roas  = attr / spend
    pct   = attr / total_sales * 100
    print(f'{ch:<10} ${spend:>11,.0f} ${attr:>11,.0f} {roas:>7.2f}x {pct:>11.1f}%')

total_media_pct = media_contrib.sum() / total_sales * 100
total_ext_pct   = ext_contrib.sum()   / total_sales * 100
total_base_pct  = GT_BASELINE * N_WEEKS / total_sales * 100
print('-' * 58)
print(f'{"Media total":<10} {"":>12} {"":>12} {"":>8} {total_media_pct:>10.1f}%')
print(f'{"External":<10} {"":>12} {"":>12} {"":>8} {total_ext_pct:>10.1f}%')
print(f'{"Baseline":<10} {"":>12} {"":>12} {"":>8} {total_base_pct:>10.1f}%')
print()
print(f'Negative sales weeks: {(df.sales < 0).sum()} / {N_WEEKS}  (should be 0)')
print(f'Holiday weeks       : {int(holiday.sum())} / {N_WEEKS}')
print(f'Comp TV shock weeks : {(comp_tv > 150).sum()} (weeks 80–94)')
print('=' * 60)
````

## Feature Engineering Pipeline

> All transformations versioned in DVC — see `dvc.yaml` in §MLOps Appendix.

| Step | Transformation | Purpose |
|------|---------------|---------|
| 1 | Geometric adstock (decay per channel) | Carryover effect |
| 2 | Hill saturation curve | Diminishing returns |
| 3 | Fourier seasonality (52w, 26w) | Annual + bi-annual cycles |
| 4 | Holiday dummies | Spike events |
| 5 | StandardScaler on external regressors | Coefficient comparability |
| 6 | Log-transform competitor GRPs | Heavy tail compression |


````python
# Feature matrix for modelling
MEDIA_FEAT_COLS = [f'{ch}_sat' for ch in MEDIA_CHANNELS]
EXT_FEAT_COLS   = ['comp_tv_std', 'cpi_std', 'rate_std', 'holiday']
FOURIER_COLS    = [c for c in df.columns if c.startswith('sin_') or c.startswith('cos_')]
ALL_FEAT_COLS   = MEDIA_FEAT_COLS + EXT_FEAT_COLS + FOURIER_COLS

X_full       = df[ALL_FEAT_COLS].values
X_media_only = df[MEDIA_FEAT_COLS + FOURIER_COLS].values
NAIVE_FEAT_COLS = [f'{ch}_spend' for ch in MEDIA_CHANNELS] + FOURIER_COLS
X_naive = df[NAIVE_FEAT_COLS].values

y = df['sales'].values

# Standardise X for coefficient comparability
scaler_full  = StandardScaler().fit(X_full)
scaler_media = StandardScaler().fit(X_media_only)
scaler_naive = StandardScaler().fit(X_naive)

Xs_full  = scaler_full.transform(X_full)
Xs_media = scaler_media.transform(X_media_only)
Xs_naive = scaler_naive.transform(X_naive)

print(f'Feature matrix shapes:')
print(f'  Full model  (media + ext + fourier) : {Xs_full.shape}')
print(f'  Adstock-only (media + fourier)      : {Xs_media.shape}')
print(f'  Naive OLS   (raw spend + fourier)   : {Xs_naive.shape}')
print(f'  (Naive includes Fourier — seasonality controlled, no adstock/Hill)')
````

## Model Fitting

### Three approaches compared:
1. **Naive OLS** — raw spend, no adstock, no externals
2. **Adstock-only** — Adstock + Hill transformed media, no externals
3. **Full Meridian-style** — Adstock + Hill + external regressors + Fourier seasonality

### Model equation:

$$
y_t = \beta_0 + \sum_c \beta_c^{\text{media}} \cdot \text{Hill}(\text{Adstock}(x_{c,t})) + \sum_k \beta_k^{\text{ext}} \cdot z_{k,t} + \sum_j \gamma_j^{\text{seasonal}} \cdot s_{j,t} + \varepsilon_t
$$

**Priors (PyMC):**
- $\beta_c^{\text{media}} \sim \text{HalfNormal}(\sigma=2000)$ — positive media response
- $\beta_k^{\text{ext}} \sim \mathcal{N}(0, 0.5)$ — informative prior around zero
- $\sigma_{\varepsilon} \sim \text{HalfNormal}(5000)$


````python
# ── 1. Naive OLS ─────────────────────────────────────────────────────────────
m_naive = LinearRegression().fit(Xs_naive, y)
y_pred_naive = m_naive.predict(Xs_naive)
r2_naive  = r2_score(y, y_pred_naive)
mape_naive = mean_absolute_percentage_error(y, y_pred_naive)

# ── 2. Adstock-only (Ridge) ───────────────────────────────────────────────────
m_adstock = Ridge(alpha=1.0).fit(Xs_media, y)
y_pred_adstock = m_adstock.predict(Xs_media)
r2_adstock  = r2_score(y, y_pred_adstock)
mape_adstock = mean_absolute_percentage_error(y, y_pred_adstock)

# ── 3. Full model ─────────────────────────────────────────────────────────────
SAMPLES_PYMC = 800
PYMC_CHAINS  = 2

if PYMC_AVAILABLE:
    print('Fitting PyMC model...')
    from sklearn.preprocessing import StandardScaler as _SS
    _yscaler = _SS()
    y_scaled = _yscaler.fit_transform(y.reshape(-1, 1)).ravel()

    with pm.Model() as mmm:
        X_data = pm.Data('X', Xs_full)

        # All priors are now dimensionless (std_y / std_X ≈ 1/1 = 1)
        beta_media   = pm.HalfNormal('beta_media',  sigma=2,   shape=len(MEDIA_FEAT_COLS))
        beta_ext     = pm.Normal('beta_ext',    mu=0, sigma=0.5, shape=len(EXT_FEAT_COLS))
        beta_fourier = pm.Normal('beta_fourier', mu=0, sigma=1,  shape=len(FOURIER_COLS))
        intercept    = pm.Normal('intercept', mu=0, sigma=1)  # y_scaled is zero-mean
        sigma        = pm.HalfNormal('sigma', sigma=1)        # y_scaled is unit-variance

        beta_all = pm.math.concatenate([beta_media, beta_ext, beta_fourier])
        mu = intercept + pm.math.dot(X_data, beta_all)
        y_obs = pm.Normal('y_obs', mu=mu, sigma=sigma, observed=y_scaled)

        trace = pm.sample(SAMPLES_PYMC, chains=PYMC_CHAINS, tune=400,
                          target_accept=0.9, progressbar=True, random_seed=RANDOM_SEED)
        ppc   = pm.sample_posterior_predictive(trace, progressbar=False)

    # Back-transform from standardised → raw $
    y_pred_sc = ppc.posterior_predictive['y_obs'].mean(('chain', 'draw')).values
    y_pred_full = _yscaler.inverse_transform(y_pred_sc.reshape(-1, 1)).ravel()

    y_pred_lo_sc = np.percentile(
        ppc.posterior_predictive['y_obs'].values.reshape(-1, N_WEEKS), 2.5, axis=0)
    y_pred_hi_sc = np.percentile(
        ppc.posterior_predictive['y_obs'].values.reshape(-1, N_WEEKS), 97.5, axis=0)
    y_pred_lo = _yscaler.inverse_transform(y_pred_lo_sc.reshape(-1, 1)).ravel()
    y_pred_hi = _yscaler.inverse_transform(y_pred_hi_sc.reshape(-1, 1)).ravel()

    ext_beta_post   = trace.posterior['beta_ext'].mean(('chain', 'draw')).values
    ext_beta_std    = trace.posterior['beta_ext'].std(('chain', 'draw')).values
    media_beta_post = trace.posterior['beta_media'].mean(('chain', 'draw')).values
    USING_PYMC = True
    print('PyMC sampling complete ✓')

else:
    print('PyMC unavailable — fitting OLS with bootstrap CIs...')
    if SM_AVAILABLE:
        ols_res = sm.OLS(y, sm.add_constant(Xs_full)).fit()
        y_pred_full = ols_res.fittedvalues
        n = len(y)
        boot_coefs = []
        for b in range(400):
            idx = np.random.choice(n, n, replace=True)
            res_b = sm.OLS(y[idx], sm.add_constant(Xs_full[idx])).fit()
            boot_coefs.append(res_b.params[1:])
        boot_coefs = np.array(boot_coefs)
        n_media = len(MEDIA_FEAT_COLS)
        n_ext   = len(EXT_FEAT_COLS)
        ext_beta_post   = boot_coefs[:, n_media:n_media+n_ext].mean(axis=0)
        ext_beta_std    = boot_coefs[:, n_media:n_media+n_ext].std(axis=0)
        media_beta_post = boot_coefs[:, :n_media].mean(axis=0)
    else:
        m_full_ridge = Ridge(alpha=1.0).fit(Xs_full, y)
        y_pred_full  = m_full_ridge.predict(Xs_full)
        coefs = m_full_ridge.coef_
        n_media = len(MEDIA_FEAT_COLS)
        n_ext   = len(EXT_FEAT_COLS)
        ext_beta_post   = coefs[n_media:n_media+n_ext]
        ext_beta_std    = np.abs(ext_beta_post) * 0.15
        media_beta_post = coefs[:n_media]

    y_pred_lo = y_pred_full * 0.93
    y_pred_hi = y_pred_full * 1.07
    USING_PYMC = False

r2_full   = r2_score(y, y_pred_full)
mape_full = mean_absolute_percentage_error(y, y_pred_full)
````

## Parameter Recovery — External Regressors

````python
EXT_NAMES = ['competitor_tv','cpi','rate_delta','holiday']
sales_std = y.std()
_raw_std = {
    'competitor_tv': df['comp_tv_std'].std(),
    'cpi':           df['cpi_std'].std(),
    'rate_delta':    df['rate_std'].std(),
    'holiday':       df['holiday'].std(),
}
GT_EXT_VEC = np.array([
    GT_BETA_EXT[k] / sales_std * _raw_std[k]
    for k in EXT_NAMES
])

if USING_PYMC:
    ext_beta_scaled = ext_beta_post          # already std_y/std_X
    ext_beta_std_sc = ext_beta_std
else:
    ext_beta_scaled = ext_beta_post / sales_std
    ext_beta_std_sc = ext_beta_std  / sales_std

param_rows = []
for i, name in enumerate(EXT_NAMES):
    gt   = GT_EXT_VEC[i]
    est  = ext_beta_scaled[i]
    s    = ext_beta_std_sc[i]
    lo95 = est - 1.96 * s
    hi95 = est + 1.96 * s
    within = '✅' if lo95 <= gt <= hi95 else '❌'
    param_rows.append({
        'Regressor':  name.replace('_', ' ').title(),
        'GT β':       round(gt,   4),
        'Est β':      round(est,  4),
        'Std':        round(s,    4),
        '95% CI Lo':  round(lo95, 4),
        '95% CI Hi':  round(hi95, 4),
        'GT in CI':   within,
        'Abs Error':  round(abs(est - gt), 4),
        'Rel Error':  f'{abs(est - gt) / (abs(gt) + 1e-9):.1%}',
    })

df_params = pd.DataFrame(param_rows)
display(df_params.style
    .background_gradient(subset=['Abs Error'], cmap='RdYlGn_r', vmin=0, vmax=0.05)
    .applymap(lambda v: 'color: #34d399; font-weight:bold' if v == '✅'
              else ('color: #ef4444; font-weight:bold' if v == '❌' else ''),
              subset=['GT in CI'])
    .format({col: '{:.4f}' for col in df_params.select_dtypes('number').columns})
    .set_properties(**{'font-size': '11px', 'text-align': 'right'})
    .set_properties(subset=['Regressor'], **{'text-align': 'left'})
    .hide(axis='index')
    .set_caption('External regressor parameter recovery — estimated vs ground truth (std_y/std_X units)')
)

EXEC_METRICS['b_comp']    = ext_beta_scaled[0]
EXEC_METRICS['b_comp_gt'] = GT_EXT_VEC[0]
EXEC_METRICS['b_cpi']     = ext_beta_scaled[1]
EXEC_METRICS['b_cpi_gt']  = GT_EXT_VEC[1]
````

````python
sanity_results = []

def chk(name, fn, category='General'):
    try:
        msg = fn()
        sanity_results.append({'Category': category, 'Check': name,
                                'Status': '✅ PASS', 'Detail': msg or ''})
    except AssertionError as e:
        sanity_results.append({'Category': category, 'Check': name,
                                'Status': '❌ FAIL', 'Detail': str(e)})
    except Exception as e:
        sanity_results.append({'Category': category, 'Check': name,
                                'Status': '⚠️ WARN', 'Detail': str(e)[:120]})

def asr(cond, msg=''): assert cond, msg

# ── 1. Data Integrity ─────────────────────────────────────────────────────────
chk('No NaN values in master DataFrame',
    lambda: asr(df.isna().sum().sum() == 0, f'{df.isna().sum().sum()} NaNs'), 'Data Integrity')
chk('Sales always positive',
    lambda: asr((df['sales'] > 0).all(), 'Negative sales found'), 'Data Integrity')
chk('Media spend always non-negative',
    lambda: asr(all((df[f'{ch}_spend'] >= 0).all() for ch in MEDIA_CHANNELS)), 'Data Integrity')
chk('Correct number of weeks',
    lambda: asr(len(df) == N_WEEKS, f'Got {len(df)} weeks'), 'Data Integrity')
chk('Index is DatetimeIndex with weekly frequency',
    lambda: asr(isinstance(df.index, pd.DatetimeIndex)), 'Data Integrity')
chk('No duplicate weeks in index',
    lambda: asr(df.index.nunique() == N_WEEKS), 'Data Integrity')

# ── 2. Adstock ────────────────────────────────────────────────────────────────
chk('All adstock decay rates in (0, 1)',
    lambda: asr(all(0 < v < 1 for v in GT_ADSTOCK.values())), 'Adstock')
chk('Adstock output monotone for constant input',
    lambda: (
        lambda a: asr(np.all(a >= 0) and a[1] > a[0])
    )(adstock(np.ones(20), 0.7)), 'Adstock')
chk('Adstock decays after impulse (step 5)',
    lambda: (
        lambda a: asr(a[6] < a[5] < a[4])
    )(adstock(np.array([0,0,0,0,10,0,0,0,0,0], dtype=float), 0.7)), 'Adstock')
chk('Hill saturation bounded in [0, 1)',
    lambda: (
        lambda h: asr(h.min() >= 0 and h.max() < 1)
    )(hill(np.linspace(0, 1000, 500), 0.8, 50)), 'Adstock')
chk('Hill saturation monotone increasing',
    lambda: (
        lambda h: asr(np.all(np.diff(h) >= 0))
    )(hill(np.linspace(0, 1000, 200), 0.8, 50)), 'Adstock')

# ── 3. Stationarity ───────────────────────────────────────────────────────────
def check_sales_stationary():
    if not SM_AVAILABLE:
        raise RuntimeError('statsmodels unavailable')
    _, p, *_ = adfuller(df['sales'].values, autolag='AIC')
    return f'ADF p={p:.4f}'

chk('ADF test on sales series', check_sales_stationary, 'Stationarity')

def check_residuals_stationary():
    if not SM_AVAILABLE:
        raise RuntimeError('statsmodels unavailable')
    resid = y - y_pred_full
    _, p, *_ = adfuller(resid, autolag='AIC')
    asr(p < 0.05, f'Residuals non-stationary (p={p:.4f})')
    return f'Residuals stationary (p={p:.4f})'

chk('ADF test on model residuals', check_residuals_stationary, 'Stationarity')

# ── 4. Multicollinearity ──────────────────────────────────────────────────────
def check_vif():
    if not SM_AVAILABLE:
        raise RuntimeError('statsmodels unavailable')
    X_check = Xs_full[:, :len(MEDIA_FEAT_COLS) + len(EXT_FEAT_COLS)]
    vifs = [variance_inflation_factor(X_check, i) for i in range(X_check.shape[1])]
    high = [(ALL_FEAT_COLS[i], round(v,1)) for i, v in enumerate(vifs) if v > 10]
    asr(len(high) == 0, f'High VIF (>10): {high}')
    return f'Max VIF={max(vifs):.2f} — OK'

chk('VIF < 10 for all media + external features', check_vif, 'Multicollinearity')
chk('Correlation between media channels < 0.90',
    lambda: (
        lambda C: asr(
            all(abs(C.loc[a,b]) < 0.90
                for a in MEDIA_CHANNELS for b in MEDIA_CHANNELS if a < b),
            'High media correlation'
        ) or 'OK'
    )(df[[f'{ch}_spend' for ch in MEDIA_CHANNELS]].corr()), 'Multicollinearity')
chk('External regressors not collinear (|r| < 0.80)',
    lambda: (
        lambda C: asr(
            all(abs(C.loc[a,b]) < 0.80
                for a in EXT_FEAT_COLS for b in EXT_FEAT_COLS if a < b),
            'Collinear external regressors'
        ) or 'OK'
    )(df[EXT_FEAT_COLS].corr()), 'Multicollinearity')

# ── 5. Model Quality ──────────────────────────────────────────────────────────
chk('Full model R² > adstock-only R²',
    lambda: asr(r2_full > r2_adstock, f'{r2_full:.4f} <= {r2_adstock:.4f}'), 'Model Quality')
chk('Full model R² > naive OLS R²',
    lambda: asr(r2_full > r2_naive, f'{r2_full:.4f} <= {r2_naive:.4f}'), 'Model Quality')
chk('Full model MAPE < 20%',
    lambda: asr(mape_full < 0.20, f'MAPE={mape_full:.2%}'), 'Model Quality')
chk('Predicted sales always positive',
    lambda: asr((y_pred_full > 0).all()), 'Model Quality')
chk('Prediction interval width > 0 everywhere',
    lambda: asr(((y_pred_hi - y_pred_lo) > 0).all()), 'Model Quality')

# ── 6. Parameter Recovery ─────────────────────────────────────────────────────
chk('Competitor TV β correctly negative',
    lambda: asr(ext_beta_post[0] < 0, f'competitor β={ext_beta_post[0]:.4f}'), 'Parameter Recovery')
chk('CPI β correctly negative',
    lambda: asr(ext_beta_post[1] < 0, f'CPI β={ext_beta_post[1]:.4f}'), 'Parameter Recovery')
chk('Holiday β correctly positive',
    lambda: asr(ext_beta_post[3] > 0, f'holiday β={ext_beta_post[3]:.4f}'), 'Parameter Recovery')
chk('GT competitor TV β within 95% CI',
    lambda: asr(
        (ext_beta_scaled[0] - 2*ext_beta_std_sc[0]) <= GT_EXT_VEC[0] <=
        (ext_beta_scaled[0] + 2*ext_beta_std_sc[0]), f'GT={GT_EXT_VEC[0]:.4f} outside CI'
    ), 'Parameter Recovery')
chk('All media betas positive',
    lambda: asr(all(v > 0 for v in media_beta_post), f'Negative media beta'), 'Parameter Recovery')

# ── 7. External / Seasonality ─────────────────────────────────────────────────
chk('Holiday weeks correctly flagged (>0 holidays in each year)',
    lambda: asr(df['holiday'].sum() >= 12), 'External Data')
chk('CPI values in realistic range [-1, 12]%',
    lambda: asr(df['cpi'].between(-1, 12).all()), 'External Data')
chk('Competitor TV spend non-negative',
    lambda: asr((df['competitor_tv'] >= 0).all()), 'External Data')
chk('Drift (competitor spike) present at week 80',
    lambda: asr(df['competitor_tv'].iloc[80:95].mean() >
                df['competitor_tv'].iloc[:80].mean() * 1.3, 'No spike'), 'External Data')
chk('Fourier terms have unit variance',
    lambda: asr(all(abs(df[c].std() - 0.707) < 0.1 for c in FOURIER_COLS[:2])), 'External Data')

df_sanity = pd.DataFrame(sanity_results)
n_pass = (df_sanity['Status'] == '✅ PASS').sum()
n_fail = (df_sanity['Status'] == '❌ FAIL').sum()
n_warn = (df_sanity['Status'] == '⚠️ WARN').sum()
EXEC_METRICS['sanity_pass']  = int(n_pass)
EXEC_METRICS['sanity_total'] = len(df_sanity)
print(f'\n══ Sanity Suite: {n_pass} PASS  {n_fail} FAIL  {n_warn} WARN ══\n')
display(df_sanity.style
    .applymap(lambda v: 'color:#16a34a;font-weight:bold' if '✅' in str(v)
              else ('color:#dc2626;font-weight:bold' if '❌' in str(v)
              else ('color:#d97706' if '⚠️' in str(v) else '')), subset=['Status'])
    .set_properties(**{'font-size':'11px'})
    .hide(axis='index'))
````

## Rolling-Origin Backtesting

````python
# Forward-Chaining Cross-Validation
TRAIN_MIN = 52    # minimum 1 year of training data
STEP      = 4     # re-fit every 4 weeks
# Week 52:  Train on weeks 1–52   → predict weeks 53–56
# Week 56:  Train on weeks 1–56   → predict weeks 57–60
# Week 60:  Train on weeks 1–60   → predict weeks 61–64
backtest_rows = []

for train_end in range(TRAIN_MIN, N_WEEKS - STEP, STEP):
    X_tr = Xs_full[:train_end]
    y_tr = y[:train_end]
    X_te = Xs_full[train_end:train_end+STEP]
    y_te = y[train_end:train_end+STEP]

    m_bt = Ridge(alpha=1.0).fit(X_tr, y_tr)
    # Ridge model is a practical constraint, the backtest loop refits ~39 times (one per step from week 52 to week 204)
    y_hat = m_bt.predict(X_te)

    mape_bt = mean_absolute_percentage_error(y_te, np.clip(y_hat, 1, None))
    for k in range(STEP):
        backtest_rows.append({
            'week': df.index[train_end + k],
            'actual': y_te[k], 'predicted': y_hat[k],
            'mape': mape_bt, 'train_size': train_end,
        })

df_bt = pd.DataFrame(backtest_rows)
overall_mape = mean_absolute_percentage_error(df_bt['actual'],
                                               np.clip(df_bt['predicted'], 1, None))
EXEC_METRICS['backtest_mape'] = overall_mape

fig_bt = go.Figure()
fig_bt.add_trace(go.Scatter(x=df_bt['week'], y=df_bt['actual'],
    name='Actual', line=dict(color=C_SKY, width=1.5)))
fig_bt.add_trace(go.Scatter(x=df_bt['week'], y=df_bt['predicted'],
    name='Backtest prediction', line=dict(color=C_ORANGE, width=1.5, dash='dash')))
fig_bt.update_layout(height=350, width=1050, template=PLOTLY_TEMPLATE,
    title=f'Rolling-origin backtest — MAPE={overall_mape:.2%}',
    yaxis_title='Weekly Sales ($)', font=dict(family=FONT_FAMILY, size=12))
fig_bt.show()
````

The Full MMM predicted unseen weekly sales within 1.57% of actuals on average — using only information available at prediction time (no future data leakage). Ridge was used as the fitting engine per iteration (same 17 features as Full MMM, MAP-equivalent estimates).

## Data Drift Detection — Competitor Spend Shock

````python
# ── Data Drift Detection — Competitor TV PSI ──────────────────────────────
COMP_SHOCK_START  = 80
COMP_SHOCK_END    = 95
COMP_BASELINE_END = COMP_SHOCK_START - 1    # 79 — clean pre-shock baseline
WINDOW            = 8
CALIBRATION_START = 20
CALIBRATION_END   = COMP_BASELINE_END - WINDOW   # weeks 20–71

def psi_kde(expected, actual, n_grid=100):
    """
    KDE-based PSI = Σᵢ (A(xᵢ) − E(xᵢ)) × ln(A(xᵢ)/E(xᵢ)) × Δx
    Unbounded — handles small windows and large distributional shifts.
    """
    kde_e = gaussian_kde(expected, bw_method='scott')
    kde_a = gaussian_kde(actual,   bw_method='scott')
    lo    = min(expected.min(), actual.min()) - 2 * expected.std()
    hi    = max(expected.max(), actual.max()) + 2 * expected.std()
    grid  = np.linspace(lo, hi, n_grid)
    dx    = grid[1] - grid[0]
    E     = np.clip(kde_e(grid), 1e-10, None)
    A     = np.clip(kde_a(grid), 1e-10, None)
    E    /= E.sum() * dx
    A    /= A.sum() * dx
    return float(np.sum((A - E) * np.log(A / E) * dx))

def first_signal(df_psi, threshold):
    hits = df_psi[df_psi['psi'] > threshold]['week']
    return hits.iloc[0] if len(hits) > 0 else None

# ── Step 1: PSI series — full timeline from week WINDOW ──────────────────
# Start from week 8 (not week 87) so the chart shows:
#   • long flat pre-shock baseline (weeks 8–79)   ← contrast period
#   • sharp spike during shock  (weeks 80–95)      ← the event
#   • clean recovery afterwards (weeks 96–207)     ← return to normal
baseline_comp = df['competitor_tv'].iloc[:COMP_BASELINE_END].values

df_psi_comp = pd.DataFrame([
    {'week': df.index[t],
     'psi':  psi_kde(baseline_comp,
                     df['competitor_tv'].iloc[t-WINDOW:t].values)}
    for t in range(WINDOW, N_WEEKS)    # full timeline — was COMP_BASELINE_END+WINDOW
])

# ── Step 2: robust threshold calibration (pre-shock windows only) ─────────
# Uses median + MAD instead of mean + std:
#   MAD is resistant to outlier windows in the calibration period
#   scale='normal' applies 1.4826 factor → MAD in σ-equivalent units
calibration_psi = np.array([
    psi_kde(baseline_comp,
            df['competitor_tv'].iloc[t-WINDOW:t].values)
    for t in range(CALIBRATION_START + WINDOW, CALIBRATION_END)
])

cal_median = np.median(calibration_psi)
cal_mad    = median_abs_deviation(calibration_psi, scale='normal')

PSI_WARNING   = cal_median + 1.5 * cal_mad
PSI_THRESHOLD = cal_median + 3.0 * cal_mad
PSI_URGENT    = cal_median + 5.0 * cal_mad

first_comp = first_signal(df_psi_comp, PSI_THRESHOLD)

print(f"Calibration PSI (pre-shock, robust)")
print(f"  median={cal_median:.3f}  MAD_σ={cal_mad:.3f}  n={len(calibration_psi)}")
print(f"\nThresholds (median + N×MAD_σ):")
print(f"  Warning  +1.5σ = {PSI_WARNING:.3f}")
print(f"  Retrain  +3.0σ = {PSI_THRESHOLD:.3f}")
print(f"  Urgent   +5.0σ = {PSI_URGENT:.3f}")
print(f"\nComp PSI range : {df_psi_comp['psi'].min():.3f} – "
      f"{df_psi_comp['psi'].max():.3f}")
print(f"Shock peak     : {df_psi_comp['psi'].max():.3f}  "
      f"({df_psi_comp['psi'].max()/PSI_THRESHOLD:.1f}x above retrain threshold)")
print(f"First signal   : {str(first_comp)[:10] if first_comp else 'NONE'}")

# ── Figure ────────────────────────────────────────────────────────────────
psi_y_max = df_psi_comp['psi'].max() * 1.12

fig = make_subplots(
    rows=2, cols=1,
    shared_xaxes=True,
    vertical_spacing=0.10,
    subplot_titles=(
        f'Competitor TV GRPs — injected shock weeks '
        f'{COMP_SHOCK_START}–{COMP_SHOCK_END} (+90 GRPs)',
        f'KDE PSI · WINDOW={WINDOW}w · '
        f'Robust thresholds median+N×MAD_σ '
        f'(median={cal_median:.2f}, MAD_σ={cal_mad:.2f})',
    ),
    row_heights=[0.35, 0.65],
)

# ── Panel 1: competitor TV GRPs ───────────────────────────────────────────
fig.add_trace(go.Scatter(
    x=df.index, y=df['competitor_tv'],
    mode='lines', name='Competitor TV GRPs',
    line=dict(color=C_PURPLE, width=1.5),
), row=1, col=1)

fig.add_vrect(
    x0=df.index[COMP_SHOCK_START].strftime('%Y-%m-%d'),
    x1=df.index[COMP_SHOCK_END].strftime('%Y-%m-%d'),
    fillcolor='rgba(239,68,68,0.15)', line_width=0,
    annotation_text='Injected shock',
    annotation_position='top left', annotation_font_size=11,
    row=1, col=1,
)

# ── Panel 2: PSI ──────────────────────────────────────────────────────────
# Zone shading — green/amber/red bands aligned to thresholds
fig.add_hrect(
    y0=0,             y1=PSI_WARNING,
    fillcolor='rgba(34,197,94,0.07)',  line_width=0, row=2, col=1)
fig.add_hrect(
    y0=PSI_WARNING,   y1=PSI_THRESHOLD,
    fillcolor='rgba(245,158,11,0.07)', line_width=0, row=2, col=1)
fig.add_hrect(
    y0=PSI_THRESHOLD, y1=1,
    fillcolor='rgba(239,68,68,0.05)',  line_width=0, row=2, col=1)

# Calibration window shading
fig.add_vrect(
    x0=df.index[CALIBRATION_START + WINDOW].strftime('%Y-%m-%d'),
    x1=df.index[CALIBRATION_END].strftime('%Y-%m-%d'),
    fillcolor='rgba(100,100,100,0.07)', line_width=0,
    annotation_text='Threshold calibration window',
    annotation_position='bottom left', annotation_font_size=9,
    row=2, col=1,
)

# Shock shading on PSI panel (mirrors panel 1 — eye connects both)
fig.add_vrect(
    x0=df.index[COMP_SHOCK_START].strftime('%Y-%m-%d'),
    x1=df.index[COMP_SHOCK_END].strftime('%Y-%m-%d'),
    fillcolor='rgba(239,68,68,0.08)', line_width=0,
    row=2, col=1,
)

# PSI line
fig.add_trace(go.Scatter(
    x=df_psi_comp['week'], y=df_psi_comp['psi'],
    mode='lines',
    name=f'PSI — Competitor TV (WINDOW={WINDOW}w)',
    line=dict(color=C_GREEN, width=2.0),
), row=2, col=1)

# Threshold lines
fig.add_hline(
    y=PSI_URGENT,
    line_dash='dash', line_color=C_NEGATIVE, line_width=1.5,
    annotation_text=f'Urgent +5σ ({PSI_URGENT:.2f})',
    annotation_font_size=8, annotation_position='top right', row=2, col=1,
)
fig.add_hline(
    y=PSI_THRESHOLD,
    line_dash='dash', line_color=C_NEGATIVE, line_width=1.5,
    annotation_text=f'Retrain +3σ ({PSI_THRESHOLD:.2f})',
    annotation_font_size=8, annotation_position='top right', row=2, col=1,
)
fig.add_hline(
    y=PSI_WARNING,
    line_dash='dot', line_color=C_ORANGE, line_width=1.5,
    annotation_text=f'Warning +1.5σ ({PSI_WARNING:.2f})',
    annotation_font_size=8, annotation_position='top right', row=2, col=1,
)

fig.update_layout(
    height=525, width=1050,
    template=PLOTLY_TEMPLATE,
    title=(
        'Data Drift Detection — Competitor TV PSI<br>'
        '<sup>'
        f'Shock weeks {COMP_SHOCK_START}–{COMP_SHOCK_END}  ·  '
        f'KDE PSI · WINDOW={WINDOW}w  ·  '
        f'Baseline: weeks 0–{COMP_BASELINE_END}  ·  '
        f'Robust thresholds: median+N×MAD_σ'
        '</sup>'
    ),
    font=dict(family=FONT_FAMILY, size=12),
    legend=dict(orientation='h', y=-0.06),
)
fig.update_yaxes(title_text='GRPs',     row=1, col=1)
fig.update_yaxes(
    title_text='PSI (KDE)',
    range=[0, 7],               row=2, col=1,
)
fig.update_xaxes(title_text='Week',     row=2, col=1)
fig.show()
````

## External Regressor Importance — Shapley-Style Variance Decomposition

````python
EXT_GROUPS = {
    'competitor_tv': [ALL_FEAT_COLS.index('comp_tv_std')],
    'cpi':           [ALL_FEAT_COLS.index('cpi_std')],
    'rate_delta':    [ALL_FEAT_COLS.index('rate_std')],
    'holiday':       [ALL_FEAT_COLS.index('holiday')],
}
MEDIA_IDX   = [ALL_FEAT_COLS.index(c) for c in MEDIA_FEAT_COLS + FOURIER_COLS]
GROUP_NAMES = list(EXT_GROUPS.keys())

def fit_r2_subgroup(feature_idx):
    """Fit Ridge on selected feature columns, return R²."""
    X_sub = Xs_full[:, feature_idx]
    m = Ridge(alpha=1.0).fit(X_sub, y)
    return m.score(X_sub, y)


n = len(GROUP_NAMES)
shapley_ext = {g: 0.0 for g in GROUP_NAMES}
fact = math.factorial

for g in GROUP_NAMES:
    others = [gg for gg in GROUP_NAMES if gg != g]
    for size in range(n):
        for coalition in combinations(others, size):
            coalition_set = list(coalition)
            weight = fact(size) * fact(n - size - 1) / fact(n)
            idx_base = MEDIA_IDX + [i for gg in coalition_set for i in EXT_GROUPS[gg]]
            idx_with = idx_base + EXT_GROUPS[g]
            r2_base = fit_r2_subgroup(idx_base)
            r2_with = fit_r2_subgroup(idx_with)
            shapley_ext[g] += weight * (r2_with - r2_base)

r2_media_only  = fit_r2_subgroup(MEDIA_IDX)
r2_full_check  = fit_r2_subgroup(list(range(Xs_full.shape[1])))

ext_shap_rows = [
    {'Component': 'Media + Seasonality (baseline)',
     'R² contribution': round(r2_media_only, 4), 'Shapley share of R²': '—'},
]
for g, v in sorted(shapley_ext.items(), key=lambda x: -x[1]):
    ext_shap_rows.append({
        'Component': g.replace('_',' ').title(),
        'R² contribution': round(v, 4),
        'Shapley share of R²': f'{v/r2_full_check:.1%}' if r2_full_check > 0 else '—'
    })
ext_shap_rows.append({'Component': 'Total (Full model)',
                       'R² contribution': round(r2_full_check, 4), 'Shapley share of R²': '100.0%'})

df_shap_ext = pd.DataFrame(ext_shap_rows)
display(df_shap_ext.style.set_properties(**{'font-size':'11px'}).hide(axis='index'))
max_shap = max(r['R² contribution'] for r in ext_shap_rows[1:-1])

fig_shap = go.Figure(go.Bar(
    x=[r['Component'] for r in ext_shap_rows[1:-1]],
    y=[r['R² contribution'] for r in ext_shap_rows[1:-1]],
    marker_color=C_PURPLE,
    text=[r['Shapley share of R²'] for r in ext_shap_rows[1:-1]],
    textposition='outside'
))
fig_shap.update_layout(height=350, width=700, template=PLOTLY_TEMPLATE,
    title='External regressor Shapley contribution to R²',
    yaxis=dict(range=[0, max_shap * 1.18]),
    yaxis_title='Marginal R² contribution', font=dict(family=FONT_FAMILY, size=12))
fig_shap.show()
````

## Bootstrap + Posterior Predictive ROAS Confidence Intervals

**Rationale:** Propagates full uncertainty (parameter + sampling) into business metrics. A point-estimate ROAS of 3.2x can become 1.8–4.8x under full uncertainty — a range that changes the budget recommendation.

**Blueprint:** Sample from posterior -> recompute mROI distribution -> report P10/P50/P90 ROAS.


````python
# ── ROAS Confidence Intervals (leave-one-channel-out) ───────────────
N_BOOT = 500
y_arr  = np.asarray(df['sales'].values, dtype=float)

roas_distributions = {ch: [] for ch in MEDIA_CHANNELS}

# media column positions in ALL_FEAT_COLS (raw, unscaled)
media_col_idx = [ALL_FEAT_COLS.index(f'{ch}_sat') for ch in MEDIA_CHANNELS]

# raw feature matrix (unscaled) — LOO zeroing happens here
X_raw_base = df[ALL_FEAT_COLS].values.copy()

if USING_PYMC:
    beta_media_draws   = trace.posterior['beta_media'].values
    beta_ext_draws     = trace.posterior['beta_ext'].values
    beta_fourier_draws = trace.posterior['beta_fourier'].values
    intercept_draws    = trace.posterior['intercept'].values

    flat_bm  = beta_media_draws.reshape(-1, len(MEDIA_FEAT_COLS))
    flat_be  = beta_ext_draws.reshape(-1, len(EXT_FEAT_COLS))
    flat_bf  = beta_fourier_draws.reshape(-1, len(FOURIER_COLS))
    flat_int = intercept_draws.reshape(-1)

    sample_idx = np.random.choice(flat_bm.shape[0], N_BOOT, replace=True)
    print(f"Drawing {N_BOOT} posterior samples for ROAS distribution...")

    # y_full computed once per sample on full scaled matrix
    y_std  = y_arr.std()
    y_mean = y_arr.mean()
    
    for s_idx in sample_idx:
        beta_all  = np.concatenate([flat_bm[s_idx], flat_be[s_idx], flat_bf[s_idx]])
        intercept = flat_int[s_idx]
    
        y_full = (intercept + Xs_full @ beta_all) * y_std + y_mean
    
        for ci, ch in enumerate(MEDIA_CHANNELS):
            X_raw_loo = X_raw_base.copy()
            X_raw_loo[:, media_col_idx[ci]] = 0.0
            Xs_loo = scaler_full.transform(X_raw_loo)
    
            y_loo       = (intercept + Xs_loo @ beta_all) * y_std + y_mean
            attributed  = (y_full - y_loo).sum()
            total_spend = df[f'{ch}_spend'].sum()
            roas_distributions[ch].append(attributed / (total_spend + 1e-6))
else:
    print(f"PyMC unavailable -- bootstrapping {N_BOOT} resamples for ROAS CI...")
    n_obs = len(y_arr)
    for b in range(N_BOOT):
        idx    = np.random.choice(n_obs, n_obs, replace=True)
        m_b    = Ridge(alpha=1.0).fit(Xs_full[idx], y_arr[idx])
        y_full = m_b.predict(Xs_full)

        for ci, ch in enumerate(MEDIA_CHANNELS):
            # Zero in raw space = true no-spend, then re-scale
            X_raw_loo = X_raw_base.copy()
            X_raw_loo[:, media_col_idx[ci]] = 0.0
            Xs_loo    = scaler_full.transform(X_raw_loo)

            y_loo      = m_b.predict(Xs_loo)
            attributed = (y_full - y_loo).sum()
            total_spend = df[f'{ch}_spend'].sum()
            roas_distributions[ch].append(attributed / (total_spend + 1e-6))

# ── Sanity check ──────────────────────────────────────────────────────────────
print("\n=== ROAS sanity check ===")
for ch in MEDIA_CHANNELS:
    dist = np.array(roas_distributions[ch])
    print(f"  {ch:10s}  median={np.median(dist):.2f}x  "
          f"p10={np.percentile(dist,10):.2f}x  p90={np.percentile(dist,90):.2f}x")

# ── Summary table ─────────────────────────────────────────────────────────────
roas_rows = []
for ch in MEDIA_CHANNELS:
    dist = np.array(roas_distributions[ch])
    roas_rows.append({
        'Channel':   ch.title(),
        'P10 ROAS':  round(np.percentile(dist, 10), 2),
        'P50 ROAS':  round(np.percentile(dist, 50), 2),
        'P90 ROAS':  round(np.percentile(dist, 90), 2),
        'Mean ROAS': round(dist.mean(), 2),
        'Std ROAS':  round(dist.std(), 2),
        'CI width':  round(np.percentile(dist, 90) - np.percentile(dist, 10), 2),
    })

df_roas_ci = pd.DataFrame(roas_rows).sort_values('P50 ROAS', ascending=False)
src = 'posterior' if USING_PYMC else 'bootstrap'
display(df_roas_ci.style
    .background_gradient(subset=['P50 ROAS'], cmap='RdYlGn', vmin=0, vmax=df_roas_ci['P50 ROAS'].max() * 1.2)
    .format('{:.2f}', subset=['P10 ROAS','P50 ROAS','P90 ROAS','Mean ROAS','Std ROAS','CI width'])
    .set_properties(**{'font-size': '11px'})
    .hide(axis='index')
    .set_caption(f'ROAS P10/P50/P90 -- full uncertainty propagation (N={N_BOOT} {src} draws)'))

# ── Violin plot ───────────────────────────────────────────────────────────────
def hex_to_rgba(hex_color, alpha=0.33):
    h = hex_color.lstrip('#')
    r, g, b = int(h[0:2], 16), int(h[2:4], 16), int(h[4:6], 16)
    return f'rgba({r},{g},{b},{alpha})'

fig_roas = go.Figure()
colors = [f'{C_SKY}', f'{C_ORANGE}', f'{C_GREEN}', f'{C_PURPLE}', f'{C_YELLOW}']
for ci, ch in enumerate(MEDIA_CHANNELS):
    fig_roas.add_trace(go.Violin(
        x=[ch.title()] * N_BOOT, y=roas_distributions[ch],
        name=ch.title(), box_visible=True, meanline_visible=True,
        fillcolor=hex_to_rgba(colors[ci], 0.33),
        line_color=colors[ci]
    ))
fig_roas.add_hline(y=1.0, line_dash='dash', line_color=f'{C_NEGATIVE}',
                   annotation_text='ROAS = 1.0 (break-even)')
fig_roas.update_layout(
    height=420, width=840, template=PLOTLY_TEMPLATE,
    title=f'ROAS distribution with full uncertainty (N={N_BOOT} {src})',
    yaxis_title='ROAS (x)', showlegend=False,
    font=dict(family=FONT_FAMILY, size=12)
)
fig_roas.show()
````

## Full ArviZ Posterior Diagnostics (ESS, R-hat, Rank Plots, PPC)

**Rationale:** MCMC validity is non-negotiable. A low ESS or R-hat > 1.01 means the sampler has not converged and coefficient estimates are unreliable — no amount of downstream analysis on a non-converged chain is valid.

**Blueprint:** `az.plot_ppc()`, `az.summary()`, `az.plot_rank()` in a dedicated diagnostics cell.


````python
# ArviZ Posterior Diagnostics

if PYMC_AVAILABLE and USING_PYMC:

    # ── Convergence Summary ───────────────────────────────────────────────────
    print("\n-- MCMC Convergence Summary --")
    summary = azs.summary(trace, var_names=['beta_ext', 'beta_media', 'intercept', 'sigma'])

    # ArviZ 1.1.0 returns numeric columns as object dtype — coerce before filtering
    for col in ['mean', 'sd', 'eti89_lb', 'eti89_ub', 'ess_bulk', 'ess_tail',
                'r_hat', 'mcse_mean', 'mcse_sd']:
        if col in summary.columns:
            summary[col] = pd.to_numeric(summary[col], errors='coerce')

    display(summary)

    bad_rhat = summary[summary['r_hat'] > 1.01]
    low_ess  = summary[summary['ess_bulk'] < 400]

    if len(bad_rhat):
        print(f"\nWARN: R-hat > 1.01: {bad_rhat.index.tolist()} -- run more samples!")
    else:
        print("\nAll R-hat <= 1.01 -- chains converged")

    if len(low_ess):
        print(f"WARN: Low bulk ESS: {low_ess.index.tolist()} -- increase tune/draws")
    else:
        print("All bulk ESS >= 400 -- sufficient effective samples")

    EXEC_METRICS['max_rhat'] = float(summary['r_hat'].max())
    EXEC_METRICS['min_ess']  = float(summary['ess_bulk'].min())

    # ── Posterior Predictive Check ────────────────────────────────────────────
    print("\n-- Posterior Predictive Check --")
    
    from arviz_plots import plot_ppc_dist, plot_rank_dist, plot_trace_dist
    
    _has_ppc = (
        'posterior_predictive' in [c.lstrip('/') for c in trace.children]
        and 'obs' in trace['posterior_predictive'].dataset
    )
    if not _has_ppc:
        with mmm:
            _ppc = pm.sample_posterior_predictive(trace, random_seed=RANDOM_SEED)
        trace.update(_ppc)
    
    plot_ppc_dist(trace)
    plt.suptitle("PPC -- model vs observed sales distribution", y=1.02)
    plt.tight_layout()
    plt.show()
    
    # ── Rank Plot ─────────────────────────────────────────────────────────────
    print("\n-- Rank plots (uniform bars = good chain mixing) --")
    plot_rank_dist(trace, var_names=['beta_ext'])
    plt.suptitle("Rank plots for beta_ext -- uniform = good mixing", y=1.02)
    plt.tight_layout()
    plt.show()
    
    # ── Trace Plot ────────────────────────────────────────────────────────────
    print("\n-- Chain traces (overlapping chains = good mixing) --")
    plot_trace_dist(trace, var_names=['beta_ext'], compact=True, combined=False)
    plt.suptitle("Chain traces for beta_ext -- two chains should overlap", y=1.02)
    plt.tight_layout()
    plt.show()

else:
    # ── Manual Gelman-Rubin R-hat (PyMC not used) ────────────────────────────
    print("\n-- Manual Gelman-Rubin R-hat (no ArviZ/PyMC) --")
    chains = trace.posterior['beta_ext'].values  # (n_chains, n_draws, n_ext)
    n_chains, n_draws, n_params = chains.shape
    rhat_manual = []
    for p in range(n_params):
        ch_means = chains[:, :, p].mean(axis=1)
        ch_vars  = chains[:, :, p].var(axis=1, ddof=1)
        B        = n_draws * ch_means.var(ddof=1)
        W        = ch_vars.mean()
        var_hat  = (n_draws - 1) / n_draws * W + B / n_draws
        rhat_manual.append(float(np.sqrt(var_hat / W)) if W > 0 else 1.0)

    ess_bulk = int(n_chains * n_draws / max(rhat_manual))
    for name, rhat in zip(EXT_NAMES, rhat_manual):
        status = 'OK' if rhat <= 1.01 else 'WARN'
        print(f"  [{status}] beta_ext[{name:15s}]  R-hat={rhat:.4f}  ESS~{ess_bulk}")

    EXEC_METRICS['max_rhat'] = max(rhat_manual)
    EXEC_METRICS['min_ess']  = ess_bulk
````

## Multicollinearity + Condition Number Monitoring

**Rationale:** Fourier seasonality terms and media adstock are often correlated (both peak in Q4), inflating coefficient standard errors and making individual channel betas unreliable even when overall R-squared is high.

**Blueprint:** VIF table + condition number + PCA fallback option.


````python
# Multicollinearity + Condition Number Monitoring
from numpy.linalg import cond
from sklearn.decomposition import PCA

# a. Variance Inflation Factor measures how much a feature's variance is "inflated" by its linear correlation with other features in the design matrix.
# VIF ≈ 1 (ideal) Feature is uncorrelated with others
# VIF = 5 Feature partially explained by others
# VIF = 10 Feature nearly collinear with others 
# VIF > 100+ Near-perfect multicollinearity

# VIF for media + external (Fourier excluded -- orthogonal by design)
vif_cols = MEDIA_FEAT_COLS + EXT_FEAT_COLS
X_vif    = Xs_full[:, [ALL_FEAT_COLS.index(c) for c in vif_cols]]
vif_rows = []

if SM_AVAILABLE:
    for i, col in enumerate(vif_cols):
        vif_val = variance_inflation_factor(X_vif, i)
        flag = 'HIGH (>10)' if vif_val > 10 else ('MOD (5-10)' if vif_val > 5 else 'OK (<5)')
        vif_rows.append({'Feature': col, 'VIF': round(vif_val, 2), 'Status': flag})
else:
    for i in range(X_vif.shape[1]):
        X_other = np.delete(X_vif, i, axis=1)
        r2_aux  = Ridge(alpha=0.01).fit(X_other, X_vif[:, i]).score(X_other, X_vif[:, i])
        vif_val = 1 / (1 - r2_aux + 1e-12)
        flag = 'HIGH' if vif_val > 10 else ('MOD' if vif_val > 5 else 'OK')
        vif_rows.append({'Feature': vif_cols[i], 'VIF': round(vif_val, 2), 'Status': flag})

df_vif  = pd.DataFrame(vif_rows).sort_values('VIF', ascending=False)
max_vif = df_vif['VIF'].max()
display(df_vif.style
    .applymap(lambda v: f'color:{C_NEGATIVE};font-weight:bold' if 'HIGH' in str(v) else
                        (f'color:{C_ORANGE}' if 'MOD' in str(v) else f'color:{C_POSITIVE}'),
              subset=['Status'])
    .set_properties(**{'font-size':'11px'}).hide(axis='index')
    .set_caption('VIF -- media + external features'))

# b. Condition number
kappa = cond(Xs_full)
# Ratio of the largest to smallest singular value of the design matrix X
# κ ≈ 1 → matrix is well-behaved, inversion is stable
# κ = 1000 → a 0.1% perturbation in data can cause ~100% error in coefficients

# Fourier columns cause numerical near-singularity at float64 precision
# cap at 1e12 for logging — anything above is a precision artifact not real collinearity
kappa_full = min(cond(Xs_full), 1e12)

# κ on media + external only -- consistent with VIF scope
kappa = cond(X_vif)   # X_vif already built above

flag_k = ('ILL-CONDITIONED (kappa>1000)' if kappa > 1000
          else 'MODERATE (100-1000)'      if kappa > 100
          else 'WELL-CONDITIONED (<100)')

print(f"\nCondition number kappa(media+ext) = {kappa:.1f}      [{flag_k}]")
print(f"Condition number kappa(full Xs)   = {kappa_full:.1f}")
print(f"Max VIF = {max_vif:.2f}")

EXEC_METRICS['condition_number']      = float(kappa)
EXEC_METRICS['condition_number_full'] = float(kappa_full)

# c. PCA fallback if kappa > 100
# PCA is applied as a pre-processing step before Ridge.
# Rotates the correlated feature space into uncorrelated principal components, keeping only enough components to explain 95% of variance. 
if kappa > 100:
    print(f"\nkappa={kappa:.0f} > 100 -- applying PCA fallback")
    X_pca_in = Xs_full[:, [ALL_FEAT_COLS.index(c) for c in MEDIA_FEAT_COLS + EXT_FEAT_COLS]]
    pca = PCA(n_components=0.95, svd_solver='full')
    X_pca = pca.fit_transform(X_pca_in)
    X_pca_full = np.column_stack([X_pca,
        Xs_full[:, [ALL_FEAT_COLS.index(c) for c in FOURIER_COLS]]])
    m_pca   = Ridge(alpha=1.0).fit(X_pca_full, y)
    r2_pca  = m_pca.score(X_pca_full, y)
    kappa_pca = cond(X_pca_full)
    print(f"  PCA components: {pca.n_components_}  variance={pca.explained_variance_ratio_.sum():.3f}")
    print(f"  PCA R2={r2_pca:.4f}  kappa_pca={kappa_pca:.1f}  (vs original {kappa:.1f})")
    EXEC_METRICS['r2_pca'] = float(r2_pca)
else:
    print(f"\nkappa={kappa:.0f} -- PCA fallback not required; Ridge sufficient")

# 6d. Correlation heatmap
corr_m = pd.DataFrame(Xs_full[:, :len(MEDIA_FEAT_COLS)], columns=MEDIA_FEAT_COLS).corr()
fig_corr = go.Figure(go.Heatmap(
    z=corr_m.values, x=MEDIA_FEAT_COLS, y=MEDIA_FEAT_COLS,
    colorscale=PALETTE_POLARITY, zmid=0, zmin=-1, zmax=1,
    text=np.round(corr_m.values, 2).astype(str), texttemplate='%{text}',
    colorbar=dict(title='r')
))
fig_corr.update_layout(height=380, width=760, template=PLOTLY_TEMPLATE,
    title='Media saturation features -- inter-channel correlation matrix',
    font=dict(family=FONT_FAMILY, size=12))
fig_corr.show()
````

## Multiverse Analysis (Specification Curve)

**Rationale:** Any single model spec (adstock decay, Hill K, Fourier order, scaler) is a researcher degree of freedom. Showing that key conclusions (competitor_tv beta < 0, macro contributions stable) hold across a universe of reasonable specs demonstrates robustness, not cherry-picking.

**Blueprint:** Define specification universe -> fit all -> plot specification curve of competitor TV beta. Python grid search.


````python
# Multiverse Analysis (Specification Curve)
SPEC_UNIVERSE = {
    'tv_decay':       [0.88, 0.92, 0.95, 0.97],
    'search_decay':   [0.25, 0.35, 0.45, 0.55],
    'tv_hill_K':      [320_000, 420_000, 520_000],
    'search_hill_K':  [80_000, 105_000, 130_000],
    'fourier_order':  [2, 3, 4],
    'scaler':         ['standard', 'robust'],
    'ridge_alpha':    [0.1, 1.0, 10.0],
}

spec_keys   = list(SPEC_UNIVERSE.keys())
spec_values = list(SPEC_UNIVERSE.values())
all_specs   = list(iproduct(*spec_values))
n_specs     = len(all_specs)
print(f"Multiverse: {n_specs} specifications")
print(f"  Grid: {' x '.join(str(len(v)) for v in spec_values)} = {n_specs}")

multiverse_results = []
for spec in all_specs:
    sd = dict(zip(spec_keys, spec))

    # ── Recompute adstock + saturation for varied channels ──────────────────
    ads_tv_s     = adstock(media_spend['tv'].values,     sd['tv_decay'])
    ads_search_s = adstock(media_spend['search'].values, sd['search_decay'])

    sat_tv_s     = hill(ads_tv_s,     GT_HILL_ALPHA['tv'],     sd['tv_hill_K'])
    sat_search_s = hill(ads_search_s, GT_HILL_ALPHA['search'], sd['search_hill_K'])

    # Display/social/email use GT params (not varied here)
    sat_display_s = df['display_sat'].values
    sat_social_s  = df['social_sat'].values
    sat_email_s   = df['email_sat'].values

    fourier_s = fourier_seasonality(N_WEEKS, periods=(52, 26), order=sd['fourier_order'])

    X_raw_s = np.column_stack([
        sat_tv_s, sat_search_s, sat_display_s, sat_social_s, sat_email_s,
        df[EXT_FEAT_COLS].values,
        fourier_s.values,
    ])

    sc_s   = StandardScaler() if sd['scaler'] == 'standard' else RobustScaler()
    X_sc_s = sc_s.fit_transform(X_raw_s)

    m_s    = Ridge(alpha=sd['ridge_alpha']).fit(X_sc_s, y)
    r2_s   = m_s.score(X_sc_s, y)
    coef_s = m_s.coef_

    # EXT block starts after 5 media cols
    N_MEDIA = 5
    multiverse_results.append({
        'tv_decay':           sd['tv_decay'],
        'search_decay':       sd['search_decay'],
        'tv_hill_K':          sd['tv_hill_K'],
        'search_hill_K':      sd['search_hill_K'],
        'fourier_order':      sd['fourier_order'],
        'scaler':             sd['scaler'],
        'ridge_alpha':        sd['ridge_alpha'],
        'R2':                 round(r2_s, 4),
        'beta_tv':            round(float(coef_s[0]), 5),
        'beta_search':        round(float(coef_s[1]), 5),
        'beta_competitor_tv': round(float(coef_s[N_MEDIA + 0]), 5),
        'beta_cpi':           round(float(coef_s[N_MEDIA + 1]), 5),
        'beta_rate_delta':    round(float(coef_s[N_MEDIA + 2]), 5),
        'beta_holiday':       round(float(coef_s[N_MEDIA + 3]), 5),
    })

df_mv = pd.DataFrame(multiverse_results)

# ── Sign stability summary ──────────────────────────────────────────────────
sign_checks = {
    'competitor_tv (expect <0)': ('beta_competitor_tv', '<'),
    'cpi           (expect <0)': ('beta_cpi',           '<'),
    'rate_delta    (expect <0)': ('beta_rate_delta',    '<'),
    'holiday       (expect >0)': ('beta_holiday',       '>'),
    'tv            (expect >0)': ('beta_tv',            '>'),
    'search        (expect >0)': ('beta_search',        '>'),
}
print(f"\nSign stability across {n_specs} specifications:")
print(f"{'Feature':<34} {'Correct sign':>12}  {'Wrong sign':>10}")
print("─" * 60)
for label, (col, direction) in sign_checks.items():
    correct = (df_mv[col] < 0).mean() if direction == '<' else (df_mv[col] > 0).mean()
    print(f"  {label:<32} {correct:>11.0%}  {1-correct:>10.0%}")

print(f"\nR2 range: [{df_mv['R2'].min():.4f}, {df_mv['R2'].max():.4f}]"
      f"  median={df_mv['R2'].median():.4f}")

# ── Store to EXEC_METRICS ───────────────────────────────────────────────────
EXEC_METRICS['multiverse_n_specs']          = n_specs
EXEC_METRICS['multiverse_comp_neg_pct']     = float((df_mv['beta_competitor_tv'] < 0).mean())
EXEC_METRICS['multiverse_holiday_pos_pct']  = float((df_mv['beta_holiday'] > 0).mean())

# ── Fig 1: Specification curve -- competitor TV ─────────────────────────────
df_mv_s = df_mv.sort_values('beta_competitor_tv').reset_index(drop=True)
colors_mv = ['#ef4444' if v < 0 else '#34d399' for v in df_mv_s['beta_competitor_tv']]

fig_mv = go.Figure(go.Bar(
    x=df_mv_s.index, y=df_mv_s['beta_competitor_tv'],
    marker_color=colors_mv,
    hovertemplate=(
        '<b>Spec %{x}</b><br>beta=%{y:.4f}<br>'
        'tv_decay=%{customdata[0]}, search_decay=%{customdata[1]}<br>'
        'tv_K=%{customdata[2]:,}, search_K=%{customdata[3]:,}<br>'
        'Fourier=%{customdata[4]}, scaler=%{customdata[5]}, alpha=%{customdata[6]}'
        '<extra></extra>'
    ),
    customdata=df_mv_s[['tv_decay','search_decay','tv_hill_K',
                         'search_hill_K','fourier_order','scaler','ridge_alpha']].values
))
fig_mv.add_hline(y=0, line_color='white', line_width=1.5)
fig_mv.add_hline(
    y=df_mv['beta_competitor_tv'].mean(), line_dash='dash', line_color='#fbbf24',
    annotation_text=f"Mean={df_mv['beta_competitor_tv'].mean():.4f}"
)
pct_neg = (df_mv['beta_competitor_tv'] < 0).mean()
fig_mv.update_layout(
    height=380, width=760, template=PLOTLY_TEMPLATE,
    title=f'Specification curve -- competitor TV beta  ({pct_neg:.0%} negative, n={n_specs})',
    xaxis_title='Specification (sorted by beta)',
    yaxis_title='beta competitor_tv',
    font=dict(family=FONT_FAMILY, size=12)
)
fig_mv.show()

# ── Fig 2: All external betas -- box plot across specs ─────────────────────
ext_beta_cols = ['beta_competitor_tv', 'beta_cpi', 'beta_rate_delta', 'beta_holiday']
ext_labels    = ['Competitor TV', 'CPI', 'Rate Delta', 'Holiday']
ext_colors_bx = [C_NEGATIVE, C_ORANGE, C_SKY, C_GREEN]

fig_box = go.Figure()
for col, label, color in zip(ext_beta_cols, ext_labels, ext_colors_bx):
    fig_box.add_trace(go.Box(
        y=df_mv[col], name=label,
        marker_color=color, line_color=color,
        boxmean='sd'
    ))
fig_box.add_hline(y=0, line_color='white', line_dash='dot', line_width=1)
fig_box.update_layout(
    height=380, width=760, template=PLOTLY_TEMPLATE,
    title=f'External regressor beta distributions across {n_specs} specifications',
    yaxis_title='Standardised beta', showlegend=False,
    font=dict(family=FONT_FAMILY, size=12)
)
fig_box.show()

# ── Fig 3: R2 histogram ─────────────────────────────────────────────────────
fig_mv_r2 = go.Figure(go.Histogram(x=df_mv['R2'], nbinsx=20, marker_color='#60a5fa'))
fig_mv_r2.add_vline(
    x=r2_full, line_dash='dash', line_color='#fbbf24',
    annotation_text=f'Base spec R2={r2_full:.4f}', 
    annotation_yshift=-30,
)
fig_mv_r2.update_layout(
    height=380, width=760, template=PLOTLY_TEMPLATE,
    title=f'R2 distribution across {n_specs} specifications',
    xaxis_title='R2', yaxis_title='Count',
    font=dict(family=FONT_FAMILY, size=12)
)
fig_mv_r2.show()

print(f"\n -- {n_specs} specifications evaluated")
````

##  Versioned Feature Store with DVC + Feast-Style Lineage

**Rationale:** Adstock decay rates and Hill K/alpha are hyperparameters — they must be versioned alongside the features they produce. A change to `GT_HILL_K['tv']` invalidates `tv_sat` everywhere and every downstream coefficient estimate. Feast-style lineage ties each feature vector to the exact parameter snapshot that generated it.

**Blueprint:** `params.yaml` -> `feature_registry.json` -> DVC stage hash -> MLflow `params_hash` tag.


````python
# Versioned Feature Store + DVC / Feast-style Lineage
import hashlib, json as _json
from datetime import datetime

PARAMS = {
    "adstock": GT_ADSTOCK,
    "hill_alpha": GT_HILL_ALPHA,
    "hill_K": GT_HILL_K,
    "fourier_periods": [52, 26],
    "fourier_order": 2,
    "scaler": "StandardScaler",
    "random_seed": RANDOM_SEED,
}

params_str  = _json.dumps(PARAMS, sort_keys=True)
params_hash = hashlib.sha256(params_str.encode()).hexdigest()[:12]
print(f"Feature param hash : {params_hash}")

feature_registry = {}
for ch in MEDIA_CHANNELS:
    feature_registry[f"{ch}_sat"] = {
        "type": "media_saturation",
        "transforms": ["geometric_adstock", "hill_saturation", "standard_scaler"],
        "adstock_decay":  GT_ADSTOCK[ch],
        "hill_alpha":     GT_HILL_ALPHA[ch],
        "hill_K":         GT_HILL_K[ch],
        "param_hash":     params_hash,
        "created_at":     datetime.utcnow().isoformat(),
        "dvc_stage":      "feature_engineering",
        "source_table":   "weekly_media_spend",
    }
for ext in ["comp_tv_std", "cpi_std", "rate_std", "holiday"]:
    feature_registry[ext] = {
        "type": "external_regressor",
        "transforms": ["standard_scaler"] if ext != "holiday" else ["binary_flag"],
        "param_hash":  params_hash,
        "created_at":  datetime.utcnow().isoformat(),
        "dvc_stage":   "ingest_external",
        "source_table": "external_macro",
    }

print("\n-- Feature Registry (Feast-style lineage) --")
print(f"{'Feature':<18} {'Type':<22} {'Transforms':<55} {'Hash'}")
print("-"*110)
for feat, meta in feature_registry.items():
    print(f"{feat:<18} {meta['type']:<22} {' -> '.join(meta['transforms']):<55} {meta['param_hash']}")

stage_fingerprint = {
    "feature_engineering": {
        "params_hash":  params_hash,
        "deps_hash":    hashlib.md5(df.to_csv().encode()).hexdigest()[:8],
        "n_features":   len(feature_registry),
        "n_rows":       N_WEEKS,
    }
}
print(f"\nDVC stage fingerprint : {stage_fingerprint}")
EXEC_METRICS['params_hash'] = params_hash
print(f"\nFeature store versioned - param hash {params_hash} covers {len(feature_registry)} features")
````
